# Custom FAQ Chatbot using Gemini 2.5 Flash

This project builds a prompt-driven FAQ chatbot for a university course using Gemini 2.5 Flash.  
The chatbot will answer course-related questions using different prompt engineering techniques.  
We will experiment with context, output formats, prompt refinement, and model parameters.  
The goal is to understand how prompts and model settings affect the chatbot's responses.

```python
from google import genai
```

```python
from google.colab import userdata
API_KEY = userdata.get("GEMINI_API_KEY")
```

```python
client = genai.Client(api_key=API_KEY)
```

```python

MODEL_NAME = "models/gemini-3.5-flash"
```

```python
prompt = """
You are a helpful FAQ chatbot for a university course.
Answer the student's question clearly and concisely.

Student question:
What is the duration of the course?
"""
```

```python
response = client.models.generate_content(
    model=MODEL_NAME,
    contents=prompt
)

print(response.text)
```

```python
course_context = """
Course Name: Artificial Intelligence and Machine Learning
Duration: 6 months
Mode: Online
Classes: Monday to Friday
Syllabus: Python, Machine Learning, Deep Learning, Generative AI
Projects: 3 practical projects
"""

contextual_prompt = f"""
You are a university course FAQ chatbot.

Use the following course information to answer the student's question:

{course_context}

Answer clearly and concisely. Do not invent information that is not provided.

Student question:
What topics are covered in the course?
"""
```

```python
response = client.models.generate_content(
    model=MODEL_NAME,
    contents=contextual_prompt
)

print(response.text)
```

```python
markdown_prompt = f"""
You are a university course FAQ chatbot.

Course information:
{course_context}

Answer the student's question using Markdown.
Use a short heading and bullet points where appropriate.

Student question:
What is included in the course?
"""
```

```python
response = client.models.generate_content(
    model=MODEL_NAME,
    contents=markdown_prompt
)

print(response.text)
```

```python
json_prompt = f"""
You are a university course FAQ chatbot.

Course information:
{course_context}

Answer the student's question in valid JSON only.

Use this structure:
{{
  "question": "...",
  "answer": "...",
  "course": "..."
}}

Student question:
What is the duration of the course?
"""
```

```python
response = client.models.generate_content(
    model=MODEL_NAME,
    contents=json_prompt
)

print(response.text)
```

```python
refined_prompt = f"""
You are a helpful university course FAQ chatbot.

Course information:
{course_context}

Answer the student's question clearly and accurately.
If relevant, provide a short example.
If useful, mention a related course resource or topic.
Do not make up information that is not present in the course information.

Student question:
How can this course help me learn Generative AI?
"""
```

```python
response = client.models.generate_content(
    model=MODEL_NAME,
    contents=refined_prompt
)

print(response.text)
```

```python
low_temperature_response = client.models.generate_content(
    model=MODEL_NAME,
    contents=refined_prompt,
    config={
        "temperature": 0.2
    }
)

print(low_temperature_response.text)
```

```python
high_temperature_response = client.models.generate_content(
    model=MODEL_NAME,
    contents=refined_prompt,
    config={
        "temperature": 0.8
    }
)

print(high_temperature_response.text)
```

```python
print("Temperature = 0.2")
print(low_temperature_response.text)

print("\n" + "*" * 90 + "\n")

print("Temperature = 0.8")
print(high_temperature_response.text)
```

```python
token_control_response = client.models.generate_content(
    model=MODEL_NAME,
    contents=refined_prompt,
    config={
        "temperature": 0.2,
        "max_output_tokens": 200
    }
)

print(token_control_response.text)
```

```python
stop_prompt = f"""
You are a university course FAQ chatbot.

Course information:
{course_context}

Answer the student's question clearly in 2-3 sentences.
End your response with the word END.

Student question:
What topics are covered in the course?
"""
```

```python
stop_sequence_response = client.models.generate_content(
    model=MODEL_NAME,
    contents=stop_prompt,
    config={
        "temperature": 0.2,
        "max_output_tokens": 150,
        "stop_sequences": ["END"]
    }
)

print(stop_sequence_response.text)
```

```python
questions = [
    "What is the duration of the course?",
    "What topics are covered?",
    "How many projects are included?",
    "What is the mode of the course?"
]

for question in questions:
    prompt = f"""
    You are a helpful university course FAQ chatbot.

    Course information:
    {course_context}

    Answer the student's question clearly and concisely.

    Student question:
    {question}
    """

    response = client.models.generate_content(
        model=MODEL_NAME,
        contents=prompt
    )

    print("Question:", question)
    print("Answer:", response.text)
    print("-" * 60)
```

```python
final_prompt = f"""
You are a university course FAQ chatbot.

Use ONLY the information provided below.

Course information:
{course_context}

Instructions:
- Answer accurately.
- Keep the answer concise.
- Do not invent information.
- Use Markdown.
- If the answer is not available in the course information, say that the information is not provided.

Student question:
What are the prerequisites for this course?
"""
```

```python
response = client.models.generate_content(
    model=MODEL_NAME,
    contents=final_prompt,
    config={
        "temperature": 0.2,
        "max_output_tokens": 150
    }
)

print(response.text)
```

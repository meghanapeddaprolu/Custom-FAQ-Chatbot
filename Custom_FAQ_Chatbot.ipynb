# Custom FAQ Chatbot using Gemini 2.5 Flash

This project builds a prompt-driven FAQ chatbot for a university course using Gemini 2.5 Flash.  
The chatbot will answer course-related questions using different prompt engineering techniques.  
We will experiment with context, output formats, prompt refinement, and model parameters.  
The goal is to understand how prompts and model settings affect the chatbot's responses.

```python
from google import genai
```

```python
from google.colab import userdata
API_KEY = userdata.get("GEMINI_API_KEY")
```

```python
client = genai.Client(api_key=API_KEY)
```

```python

MODEL_NAME = "models/gemini-3.6-flash"
```

```python
prompt = """
You are a helpful FAQ chatbot for a university course.
Answer the student's question clearly and concisely.

Student question:
What is the duration of the course?
"""
```

```python
response = client.models.generate_content(
    model=MODEL_NAME,
    contents=prompt
)

print(response.text)
```

```python
course_context = """
Course Name: Artificial Intelligence and Machine Learning
Duration: 6 months
Mode: Online
Classes: Monday to Friday
Syllabus: Python, Machine Learning, Deep Learning, Generative AI
Projects: 3 practical projects
"""

contextual_prompt = f"""
You are a university course FAQ chatbot.

Use the following course information to answer the student's question:

{course_context}

Answer clearly and concisely. Do not invent information that is not provided.

Student question:
What topics are covered in the course?
"""
```

```python
response = client.models.generate_content(
    model=MODEL_NAME,
    contents=contextual_prompt
)

print(response.text)
```

```python
markdown_prompt = f"""
You are a university course FAQ chatbot.

Course information:
{course_context}

Answer the student's question using Markdown.
Use a short heading and bullet points where appropriate.

Student question:
What is included in the course?
"""
```

```python
response = client.models.generate_content(
    model=MODEL_NAME,
    contents=markdown_prompt
)

print(response.text)
```

```python
json_prompt = f"""
You are a university course FAQ chatbot.

Course information:
{course_context}

Answer the student's question in valid JSON only.

Use this structure:
{{
  "question": "...",
  "answer": "...",
  "course": "..."
}}

Student question:
What is the duration of the course?
"""
```

```python
response = client.models.generate_content(
    model=MODEL_NAME,
    contents=json_prompt
)

print(response.text)
```

```python
refined_prompt = f"""
You are a helpful university course FAQ chatbot.

Course information:
{course_context}

Answer the student's question clearly and accurately.
If relevant, provide a short example.
If useful, mention a related course resource or topic.
Do not make up information that is not present in the course information.

Student question:
How can this course help me learn Generative AI?
"""
```

```python
response = client.models.generate_content(
    model=MODEL_NAME,
    contents=refined_prompt
)

print(response.text)
```
